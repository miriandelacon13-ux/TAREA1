# 3.1 Pregunta 1: ¿Qué departamentos concentran el mayor monto recaudado?
### Pregunta 1: ¿Qué departamentos concentran el mayor monto recaudado durante 2025?

```python
tabla1 = (
    df.groupby("DEPARTAMENTO_EJECUTORA_NOMBRE", as_index=False)["MONTO_RECAUDADO"]
      .sum()
      .sort_values("MONTO_RECAUDADO", ascending=False)
)

tabla1.head(10)
```

Agrupé los registros por departamento y sumé el monto recaudado para conocer cuáles concentran los mayores ingresos durante el año 2025. Luego ordené los resultados de mayor a menor y mostré los diez primeros departamentos.

Los resultados muestran que Lima concentra ampliamente el mayor monto recaudado durante 2025. Después aparecen La Libertad, Tacna, Arequipa y Piura, aunque con montos considerablemente menores.

3.2 Pregunta 2: ¿Qué entidades concentran la recaudación dentro de los departamentos con mayores ingresos?

```python
top5_departamentos = tabla1.head(5)["DEPARTAMENTO_EJECUTORA_NOMBRE"]

top5_departamentos
```

Seleccioné los cinco departamentos con mayor monto recaudado según los resultados de la primera pregunta, para analizar qué entidades explican principalmente esos ingresos.

```python
tabla2 = (
    df[df["DEPARTAMENTO_EJECUTORA_NOMBRE"].isin(top5_departamentos)]
      .groupby(
          ["DEPARTAMENTO_EJECUTORA_NOMBRE", "EJECUTORA_NOMBRE"],
          as_index=False
      )["MONTO_RECAUDADO"]
      .sum()
      .sort_values(
          ["DEPARTAMENTO_EJECUTORA_NOMBRE", "MONTO_RECAUDADO"],
          ascending=[True, False]
      )
)

tabla2
```

Filtré la base para conservar únicamente los cinco departamentos con mayor recaudación y luego agrupé la información por departamento y entidad ejecutora. Esto permite identificar qué organismos aportan los mayores montos recaudados dentro de cada departamento.

```python
tabla2_top3 = (
    tabla2.groupby("DEPARTAMENTO_EJECUTORA_NOMBRE")
          .head(3)
          .reset_index(drop=True)
)

tabla2_top3
```

Para facilitar la lectura, seleccioné las tres entidades con mayor monto recaudado dentro de cada uno de los cinco departamentos analizados.

La tabla permite observar que la recaudación no se distribuye de manera uniforme entre las entidades. En varios departamentos, una pequeña cantidad de organismos concentra una parte importante del monto recaudado.

### Pregunta 3: ¿Cómo evolucionó mensualmente la recaudación en los cinco departamentos con mayores ingresos?

```python
tabla3 = pd.pivot_table(
    df[df["DEPARTAMENTO_EJECUTORA_NOMBRE"].isin(top5_departamentos)],
    index="MES_DOC",
    columns="DEPARTAMENTO_EJECUTORA_NOMBRE",
    values="MONTO_RECAUDADO",
    aggfunc="sum",
    fill_value=0
)

tabla3
```

Construí una tabla dinámica para comparar el monto recaudado por mes entre los cinco departamentos con mayor recaudación total. Utilicé los meses como filas y los departamentos como columnas, sumando el monto recaudado en cada caso.

La recaudación presenta variaciones importantes entre los meses. Lima mantiene los mayores montos durante todo el periodo analizado, mientras que en los demás departamentos se observan meses específicos con incrementos más elevados, como enero en Tacna y Piura.

```python
meses = {
    1: "Enero",
    2: "Febrero",
    3: "Marzo",
    4: "Abril",
    5: "Mayo",
    6: "Junio",
    7: "Julio",
    8: "Agosto",
    9: "Septiembre",
    10: "Octubre",
    11: "Noviembre",
    12: "Diciembre"
}

tabla3.index = tabla3.index.map(meses)

tabla3
```

Reemplacé los números de los meses por sus nombres para que la tabla sea más fácil de interpretar.

## 4. Cruce con una segunda fuente

Como segunda fuente se utilizaron las estimaciones de población por departamento del Instituto Nacional de Estadística e Informática (INEI) para el año 2025. Esta información permitirá relacionar el monto recaudado con la cantidad de habitantes de cada departamento.

```python
poblacion_raw = pd.read_excel(
    "poblacion_departamentos_2025.xlsx",
    sheet_name="2024-2026",
    header=None
)

poblacion_raw.head(10)
```

Cargué la hoja correspondiente al periodo 2024-2026 del archivo del INEI para trabajar con la población estimada por departamento del año 2025.

Seleccionamos solo las columnas necesarias

En este Excel, la columna 1 corresponde al departamento y la columna 5 contiene la población total de 2025.

```python
poblacion = poblacion_raw.iloc[5:, [1, 5]].copy()

poblacion.columns = [
    "DEPARTAMENTO",
    "POBLACION_2025"
]

poblacion.head()
```

Seleccioné únicamente el nombre del departamento y la población total estimada para 2025, porque son las variables necesarias para realizar el cruce con la base de ingresos.

Eliminamos la fila nacional

El Excel también contiene una fila llamada Perú, pero nosotros necesitamos departamentos:

```python
poblacion = poblacion[
    poblacion["DEPARTAMENTO"] != "Perú"
].copy()

poblacion.head()
```

Revisamos la base resultante

```python
poblacion.shape
```

```python
poblacion.head(10)
```

```python
import unicodedata

def quitar_tildes(texto):
    return "".join(
        caracter
        for caracter in unicodedata.normalize("NFD", texto)
        if unicodedata.category(caracter) != "Mn"
    )
```

Creé una función para eliminar las tildes de los nombres de los departamentos. Esto permitirá que nombres como JUNIN y Junín sean considerados iguales durante el cruce.

```python
poblacion = poblacion.dropna(subset=["DEPARTAMENTO"]).copy()

poblacion = poblacion[
    poblacion["DEPARTAMENTO"] != "Perú"
].copy()
```

Eliminé las filas vacías y la fila correspondiente al total nacional, ya que para el análisis solo necesito registros a nivel de departamento.

```python
poblacion["DEPARTAMENTO_CLAVE"] = (
    poblacion["DEPARTAMENTO"]
    .str.strip()
    .str.upper()
    .apply(quitar_tildes)
)
```

```python
recaudacion_departamento = (
    df.groupby(
        "DEPARTAMENTO_EJECUTORA_NOMBRE",
        as_index=False
    )["MONTO_RECAUDADO"]
    .sum()
)

recaudacion_departamento.head()
```

Agrupé la base de ingresos por departamento y sumé el monto recaudado. Esto permite tener una sola observación por departamento antes de realizar el cruce con la población.

```python
recaudacion_departamento["DEPARTAMENTO_CLAVE"] = (
    recaudacion_departamento["DEPARTAMENTO_EJECUTORA_NOMBRE"]
    .str.strip()
    .str.upper()
    .apply(quitar_tildes)
)
```

Estandaricé los nombres de los departamentos convirtiéndolos a mayúsculas, eliminando espacios adicionales y retirando las tildes. De esta manera, ambas bases tienen una llave comparable.

### Llave del cruce

La llave utilizada para unir ambas bases es el departamento. En la base de ingresos, la variable original es `DEPARTAMENTO_EJECUTORA_NOMBRE`, mientras que en la base del INEI se denomina `DEPARTAMENTO`.

Para evitar problemas por diferencias en mayúsculas y tildes, se creó en ambas bases una variable estandarizada llamada `DEPARTAMENTO_CLAVE`.

```python
cruce = recaudacion_departamento.merge(
    poblacion[
        [
            "DEPARTAMENTO",
            "POBLACION_2025",
            "DEPARTAMENTO_CLAVE"
        ]
    ],
    on="DEPARTAMENTO_CLAVE",
    how="left",
    indicator=True
)

cruce.head()
```

Realicé un cruce de tipo left para conservar todos los departamentos de la base de ingresos y agregar la población estimada del INEI cuando exista coincidencia. También utilicé un indicador para verificar el resultado del cruce.

```python
cruce["_merge"].value_counts()
```

Los 12 departamentos presentes en la base de ingresos encontraron correspondencia en la base del INEI. No quedaron registros sin cruzar después de estandarizar los nombres de los departamentos.

Creamos la variable que necesita ambas fuentes

Ahora vamos a responder nuestra pregunta:

¿Cuánto ingreso recaudado corresponde por habitante en cada departamento?

```python
cruce["POBLACION_2025"] = pd.to_numeric(
    cruce["POBLACION_2025"]
)
```

```python
cruce["RECAUDACION_POR_HABITANTE"] = (
    cruce["MONTO_RECAUDADO"]
    / cruce["POBLACION_2025"]
)
```

Calculé la recaudación por habitante dividiendo el monto total recaudado de cada departamento entre su población estimada para 2025. Este indicador solo puede obtenerse al combinar la información de las dos fuentes.

## Tabla final

```python
tabla_cruce = (
    cruce[
        [
            "DEPARTAMENTO_EJECUTORA_NOMBRE",
            "MONTO_RECAUDADO",
            "POBLACION_2025",
            "RECAUDACION_POR_HABITANTE"
        ]
    ]
    .sort_values(
        "RECAUDACION_POR_HABITANTE",
        ascending=False
    )
    .reset_index(drop=True)
)

tabla_cruce
```

Organicé la tabla final de mayor a menor según la recaudación por habitante. De esta manera se puede comparar la recaudación de los departamentos teniendo en cuenta el tamaño de su población.

```python
tabla_cruce["RECAUDACION_POR_HABITANTE"] = (
    tabla_cruce["RECAUDACION_POR_HABITANTE"].round(2)
)

tabla_cruce
```

Redondeé la recaudación por habitante a dos decimales para facilitar la lectura e interpretación de los resultados.
